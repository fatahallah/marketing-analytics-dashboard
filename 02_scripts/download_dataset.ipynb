```python
import os
import pandas as pd
import numpy as np

# 1. تحديد مسارات الملفات
current_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in locals() else os.getcwd()
raw_path = os.path.abspath(os.path.join(current_dir, "..", "01_data", "raw", "marketing_sales_data.csv"))
processed_dir = os.path.abspath(os.path.join(current_dir, "..", "01_data", "processed"))
os.makedirs(processed_dir, exist_ok=True)

processed_path = os.path.join(processed_dir, "marketing_cleaned.csv")

# 2. قراءة البيانات الخام
df = pd.read_csv(raw_path)
print("جاري تنظيف وتجهيز البيانات...")

# 3. معالجة القيم المفقودة
df['Income'] = df['Income'].fillna(df['Income'].median())

# 4. الهندسة الحسابية للأعمدة (Feature Engineering)
df['Age'] = 2026 - df['Year_Birth']
df['Children'] = df['Kidhome'] + df['Teenhome']

# إجمالي الإنفاق التسويقي والمبيعات
spend_cols = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']
df['Total_Spend'] = df[spend_cols].sum(axis=1)

# إجمالي العمليات الشرائية
purchase_cols = ['NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases']
df['Total_Purchases'] = df[purchase_cols].sum(axis=1)

# إجمالي الاستجابة للحملات الإعلانية
cmp_cols = ['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5', 'Response']
df['Total_Accepted_Cmp'] = df[cmp_cols].sum(axis=1)

# استبعاد الاعمدة غير المؤثرة (Z_CostContact و Z_Revenue ثوابت)
df.drop(columns=['Z_CostContact', 'Z_Revenue'], inplace=True, errors='ignore')

# 5. حفظ الملف النظيف
df.to_csv(processed_path, index=False)

print(f"تم تنظيف البيانات وحفظها بنجاح في: {processed_path}")
print(f"الأبعاد الجديدة للبيانات: {df.shape[0]} صف و {df.shape[1]} عمود.")
```

```python
import os
import sqlite3
import pandas as pd

# 1. تحديد المسارات
current_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in locals() else os.getcwd()
processed_path = os.path.abspath(os.path.join(current_dir, "..", "01_data", "processed", "marketing_cleaned.csv"))
sql_dir = os.path.abspath(os.path.join(current_dir, "..", "03_sql"))
os.makedirs(sql_dir, exist_ok=True)

db_path = os.path.join(sql_dir, "marketing_analytics.db")

# 2. تحميل البيانات النظيفة إلى قاعدة بيانات SQLite
df = pd.read_csv(processed_path)
conn = sqlite3.connect(db_path)

df.to_sql("marketing_campaigns", conn, if_exists="replace", index=False)
print("تم إنشاء قاعدة البيانات SQLite ونقل البيانات إليها بنجاح!")

# 3. حفظ ملف schema.sql
schema_content = """-- Schema for Digital Marketing & Ecommerce Performance
CREATE TABLE IF NOT EXISTS marketing_campaigns (
    ID INTEGER PRIMARY KEY,
    Year_Birth INTEGER,
    Education TEXT,
    Marital_Status TEXT,
    Income REAL,
    Kidhome INTEGER,
    Teenhome INTEGER,
    Dt_Customer TEXT,
    Recency INTEGER,
    MntWines REAL,
    MntFruits REAL,
    MntMeatProducts REAL,
    MntFishProducts REAL,
    MntSweetProducts REAL,
    MntGoldProds REAL,
    NumDealsPurchases INTEGER,
    NumWebPurchases INTEGER,
    NumCatalogPurchases INTEGER,
    NumStorePurchases INTEGER,
    NumWebVisitsMonth INTEGER,
    AcceptedCmp3 INTEGER,
    AcceptedCmp4 INTEGER,
    AcceptedCmp5 INTEGER,
    AcceptedCmp1 INTEGER,
    AcceptedCmp2 INTEGER,
    Complain INTEGER,
    Response INTEGER,
    Age INTEGER,
    Children INTEGER,
    Total_Spend REAL,
    Total_Purchases INTEGER,
    Total_Accepted_Cmp INTEGER
);
"""

with open(os.path.join(sql_dir, "schema.sql"), "w", encoding="utf-8") as f:
    f.write(schema_content)

# 4. تشغيل استعلامات SQL التحليلية وحفظ النتائج
queries = {
    "1_Category_Performance": """
        SELECT 
            SUM(MntWines) AS Total_Wines,
            SUM(MntFruits) AS Total_Fruits,
            SUM(MntMeatProducts) AS Total_Meat,
            SUM(MntFishProducts) AS Total_Fish,
            SUM(MntSweetProducts) AS Total_Sweets,
            SUM(MntGoldProds) AS Total_Gold
        FROM marketing_campaigns;
    """,
    "2_Channel_Breakdown": """
        SELECT 
            SUM(NumWebPurchases) AS Total_Web_Purchases,
            SUM(NumCatalogPurchases) AS Total_Catalog_Purchases,
            SUM(NumStorePurchases) AS Total_Store_Purchases
        FROM marketing_campaigns;
    """,
    "3_Campaign_Acceptance_Rates": """
        SELECT 
            SUM(AcceptedCmp1) AS Cmp1_Conversions,
            SUM(AcceptedCmp2) AS Cmp2_Conversions,
            SUM(AcceptedCmp3) AS Cmp3_Conversions,
            SUM(AcceptedCmp4) AS Cmp4_Conversions,
            SUM(AcceptedCmp5) AS Cmp5_Conversions,
            SUM(Response) AS Final_Campaign_Conversions
        FROM marketing_campaigns;
    """
}

# كتابة الاستعلامات في ملف analytics_queries.sql
with open(os.path.join(sql_dir, "analytics_queries.sql"), "w", encoding="utf-8") as f:
    for name, q in queries.items():
        f.write(f"-- {name}\n{q.strip()}\n\n")

# طباعة نتائج الاستعلامات
print("\n--- نتائج استعلامات SQL ---")
for name, q in queries.items():
    print(f"\n[QUERY: {name}]")
    res = pd.read_sql_query(q, conn)
    print(res)

conn.close()
print(f"\nتم حفظ كافة ملحقات SQL في المجلد: {sql_dir}")
```

```python
import os
import json
import sqlite3
import pandas as pd

# 1. تحديد المسارات
current_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in locals() else os.getcwd()
db_path = os.path.abspath(os.path.join(current_dir, "..", "03_sql", "marketing_analytics.db"))

react_data_dir = os.path.abspath(os.path.join(current_dir, "..", "05_interactive_web", "src", "data"))
os.makedirs(react_data_dir, exist_ok=True)

json_output_path = os.path.join(react_data_dir, "marketingData.json")

# 2. اتصال بقاعدة البيانات وسحب البيانات المجمعة
conn = sqlite3.connect(db_path)

# أ) أداء الفئات (Category Spend)
df_cat = pd.read_sql_query("""
    SELECT 'Wines' AS category, SUM(MntWines) AS spend FROM marketing_campaigns
    UNION ALL SELECT 'Meat', SUM(MntMeatProducts) FROM marketing_campaigns
    UNION ALL SELECT 'Gold', SUM(MntGoldProds) FROM marketing_campaigns
    UNION ALL SELECT 'Fish', SUM(MntFishProducts) FROM marketing_campaigns
    UNION ALL SELECT 'Sweets', SUM(MntSweetProducts) FROM marketing_campaigns
    UNION ALL SELECT 'Fruits', SUM(MntFruits) FROM marketing_campaigns;
""", conn)

# ب) توزيع القنوات (Channel Breakdown)
df_channels = pd.read_sql_query("""
    SELECT 'Store' AS channel, SUM(NumStorePurchases) AS purchases FROM marketing_campaigns
    UNION ALL SELECT 'Web', SUM(NumWebPurchases) FROM marketing_campaigns
    UNION ALL SELECT 'Catalog', SUM(NumCatalogPurchases) FROM marketing_campaigns;
""", conn)

# ج) أداء الحملات التسويقية (Campaign Conversions)
df_campaigns = pd.read_sql_query("""
    SELECT 'Campaign 1' AS campaign, SUM(AcceptedCmp1) AS conversions FROM marketing_campaigns
    UNION ALL SELECT 'Campaign 2', SUM(AcceptedCmp2) FROM marketing_campaigns
    UNION ALL SELECT 'Campaign 3', SUM(AcceptedCmp3) FROM marketing_campaigns
    UNION ALL SELECT 'Campaign 4', SUM(AcceptedCmp4) FROM marketing_campaigns
    UNION ALL SELECT 'Campaign 5', SUM(AcceptedCmp5) FROM marketing_campaigns
    UNION ALL SELECT 'Final Campaign', SUM(Response) FROM marketing_campaigns;
""", conn)

# د) المؤشرات الرئيسية (Summary KPIs)
summary_kpis = {
    "total_revenue": int(pd.read_sql_query("SELECT SUM(Total_Spend) AS total FROM marketing_campaigns;", conn)['total'][0]),
    "total_customers": int(pd.read_sql_query("SELECT COUNT(*) AS total FROM marketing_campaigns;", conn)['total'][0]),
    "avg_income": float(round(pd.read_sql_query("SELECT AVG(Income) AS avg FROM marketing_campaigns;", conn)['avg'][0], 2)),
    "total_purchases": int(pd.read_sql_query("SELECT SUM(Total_Purchases) AS total FROM marketing_campaigns;", conn)['total'][0])
}

conn.close()

# 3. تجميع البيانات وتصديرها كـ JSON
final_payload = {
    "summary": summary_kpis,
    "categories": df_cat.to_dict(orient="records"),
    "channels": df_channels.to_dict(orient="records"),
    "campaigns": df_campaigns.to_dict(orient="records")
}

with open(json_output_path, "w", encoding="utf-8") as f:
    json.dump(final_payload, f, indent=4, ensure_ascii=False)

print(f"تم تصدير ملف JSON للتطبيقات المتقدمة بنجاح في المسار:\n{json_output_path}")
```

```python
import os
import sqlite3
import pandas as pd

# 1. تحديد المسارات
current_dir = os.path.dirname(os.path.abspath(__file__)) if '__file__' in locals() else os.getcwd()
processed_path = os.path.abspath(os.path.join(current_dir, "..", "01_data", "processed", "marketing_cleaned.csv"))
sql_dir = os.path.abspath(os.path.join(current_dir, "..", "03_sql"))
db_path = os.path.join(sql_dir, "marketing_analytics.db")

# 2. قراءة البيانات النظيفة
df = pd.read_csv(processed_path)
conn = sqlite3.connect(db_path)

# حفظ الجدول الرئيسي
df.to_sql("marketing_campaigns", conn, if_exists="replace", index=False)

# 3. إنشاء جدول المقاييس الجاهزة (KPIs Summary) لـ Power BI مباشرة
kpi_query = """
CREATE TABLE IF NOT EXISTS powerbi_kpi_summary AS
SELECT 
    COUNT(ID) AS Total_Customers,
    ROUND(SUM(Total_Spend), 2) AS Total_Revenue,
    SUM(Total_Purchases) AS Total_Purchases,
    ROUND(AVG(Income), 2) AS Avg_Customer_Income,
    SUM(Total_Accepted_Cmp) AS Total_Conversions,
    ROUND(CAST(SUM(Response) AS FLOAT) / COUNT(ID) * 100, 2) AS Final_Conversion_Rate_Pct
FROM marketing_campaigns;
"""

# إعادة إنشاء جدول المقاييس المجمعة
conn.execute("DROP TABLE IF EXISTS powerbi_kpi_summary;")
conn.execute(kpi_query)
conn.commit()

print("تم تجهيز قاعدة البيانات SQLite وجداول Power BI المجمعة بنجاح!")
print(f"مسار ملف قاعدة البيانات:\n{db_path}")

conn.close()
```
